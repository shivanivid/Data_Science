```python
# Let's create a copy of the dataset

df_copy = df.copy()
```

```python
df_copy.drop(['Player'],axis = 1,inplace =True)
```

```python
df_copy
```

```python
# Standardization

# Age(70)  , Salary(700000)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler()
```

```python
df_scaled = sc.fit_transform(df_copy) # fit - learn , transform - apply
```

```python
df_scaled
```

```python
df_scaled = pd.DataFrame(df_scaled,columns = df_copy.columns)
```

```python
df_scaled
```

```python
df
```

```python
from sklearn.cluster import KMeans
```

```python
k_values = [2,3,4,5,6,7]
```

```python
# elbow method

ssd = []

for k in k_values:
  km = KMeans(n_clusters=k,max_iter=150, random_state = 32)
  km.fit(df_scaled)
  ssd.append(km.inertia_)
```

```python
plt.plot(k_values,ssd) # Sharp bend
```

```python
# elbow method - 3 or 4
```

```python
# Silhouette score
```

image.png

```python
from sklearn.metrics import  silhouette_score

k_values = range(2,8)

silhouette_scores = []

for k in k_values:
  kmeans = KMeans(n_clusters = k,random_state = 32 )
  kmeans.fit(df_scaled)
  silhouette_avg = silhouette_score(df_scaled,kmeans.labels_)
  silhouette_scores.append(silhouette_avg)

plt.subplot(1,2,2)

plt.plot(k_values,silhouette_scores,marker = 'o',color = "green")





```

```python
# Final value - 4

kmodel = KMeans(n_clusters = 4,max_iter = 150,random_state = 42)
kmodel.fit(df_scaled)
```

```python
print(kmodel.labels_)
```

```python
df['Cluster_ID'] = kmodel.labels_
```
