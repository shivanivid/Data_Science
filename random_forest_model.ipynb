```python
# pd.to_numeric()

# any columns is converted into numeric data

# if there is any string value it can also be handled

# 'abc' - cannot convert into numeric value

# so we use parameter errors='coerce'
# if there is any atring value which cannot be converted into numeric it
# converts the value into a null value
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'])
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'],errors='coerce')
```

```python
df.isnull().sum()
```

```python
# handle the null values

# -- drop the values
# --  fill with specific value
# -- fill with mean,median or mode
```

```python
df['TotalCharges'] = df['TotalCharges'].fillna(df['MonthlyCharges']*df['tenure'])
```

```python
df.drop(columns=['customerID'],inplace=True)
```

Encoding - To convert string or object data into numeric so that we can use it with the algo.

Label Encoding - yes-1, no-0, The changes are done in the place in the same col.


```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
for col in df.columns:
  if df[col].dtype == 'object':
    df[col] = le.fit_transform(df[col])
```

```python
df.info()
```

```python
# dividing the data into independent and dependent col

x = df.drop(columns=['Churn'])
y = df['Churn']
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2, random_state=7)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
RF_model = RandomForestClassifier(n_estimators=300, oob_score=True)
```

```python
RF_model.fit(x_train,y_train)
```

```python
y_pred = RF_model.predict(x_test)
```

```python
from sklearn.metrics import *
```

```python
accuracy_score(y_pred,y_test)
```

```python
# TAKING different values for the same parameter

n = [50,100,200,300,500,700]

for i in n:
  model = RandomForestClassifier(n_estimators=i)
  model.fit(x_train,y_train)
  y_pred = model.predict(x_test)
  a = accuracy_score(y_test,y_pred)
  print(a,i)
```
